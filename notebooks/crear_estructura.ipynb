## Creación automatizada de la estructura del proyecto

En esta sección se prepara de forma automática la estructura base del proyecto con el objetivo de garantizar una organización consistente de archivos, facilitar la colaboración entre los integrantes del equipo y asegurar la reproducibilidad del análisis.

### Importación de dependencias

Inicialmente se importan las librerías necesarias para trabajar con rutas de archivos, ejecutar comandos del sistema y gestionar las dependencias del proyecto:

- `pathlib.Path`: permite manipular rutas y directorios de forma independiente del sistema operativo.
- `sys`: proporciona acceso al intérprete de Python que ejecuta el Notebook.
- `subprocess`: permite ejecutar comandos del sistema desde Python.
- `importlib.metadata`: se utiliza para consultar las versiones instaladas de los paquetes y verificar que todas las dependencias requeridas estén disponibles.

### Definición del directorio raíz del proyecto

Posteriormente se identifica la raíz del proyecto mediante el directorio de trabajo actual del Notebook. Esto permite construir rutas relativas de forma consistente sin depender de ubicaciones absolutas específicas de cada equipo.

> **Importante:** el directorio de trabajo del kernel de Jupyter no siempre coincide con la ubicación física del Notebook. Por este motivo, el Notebook debe ejecutarse utilizando la raíz del proyecto como directorio de trabajo.

### Generación de la estructura de carpetas

Se define una estructura estándar de directorios para separar adecuadamente los distintos componentes del proyecto:

- `data/raw`: datos originales sin procesar.
- `data/processed`: datos transformados o preparados para el análisis.
- `data/external`: datos obtenidos de fuentes externas.
- `notebooks`: Notebooks utilizados durante el proceso de exploración y análisis.
- `src`: código fuente reutilizable del proyecto.
- `reports/figures`: gráficos y visualizaciones generadas.
- `reports/outputs`: resultados finales, tablas y reportes.
- `models`: modelos o artefactos producidos durante el análisis.

Posteriormente, se crea automáticamente cada una de estas carpetas. Para ello se utilizan las opciones:

- `parents=True`: permite crear directorios padre que aún no existan.
- `exist_ok=True`: evita generar errores cuando un directorio ya se encuentra creado.

De esta forma, el procedimiento puede ejecutarse múltiples veces sin afectar la estructura existente.

### Preparación de la estructura para Git

Finalmente, se genera un archivo `.gitkeep` dentro de cada directorio creado.

El propósito de estos archivos es permitir que Git rastree carpetas vacías, ya que por diseño Git únicamente versiona archivos y no directorios vacíos. Gracias a esto, todos los integrantes del equipo obtendrán exactamente la misma estructura de carpetas al clonar el repositorio, incluso antes de que existan archivos de datos o resultados dentro de ellas.

Esta práctica mejora la organización del proyecto y facilita la colaboración entre los miembros del equipo al mantener una estructura de trabajo homogénea desde el primer commit.

````python

from pathlib import Path    # Path nos permite trabajar con rutas y directorios.
import sys                  # Proporciona acceso al intérprete de Python
import subprocess           # Ejecuta comandos del sistema desde Python
from importlib.metadata import version, PackageNotFoundError    # Nos permite consultar 
# las versiones intaladas de nuestras dependencias y capturar un error cuando un paquete 
# no está instalado.


# Directorio principal del proyecto 
PROJECT_ROOT = Path.cwd().parent   # Obtiene el directorio de trabajo actual del Notebook

# Recordar que Path.cwd() utiliza el directorio de trabajo del kernel, que no necesariamente deberá
# coincidir con la ubicación física del Notebook. Por lo que se deberá ejecutar el Notebook en la
# raíz del proyecto como directorio de trabajo.

# Estructura de directorios que se desean crear para el proyecto
folders = [
    "data/raw",
    "data/processed",
    "data/external",
    "notebooks",
    "src",
    "reports/figures",
    "reports/outputs",
    "models"
]

# Crear directorios
for folder in folders:
    path = PROJECT_ROOT / folder
    path.mkdir(parents=True, exist_ok=True)    # mkdir() >> Crea el directorio 
    print(f"Directorio creado: {path}")        # parents=True >> Crea los directorios superiores 
                                               # exist_ok=True >> Evita errores si la carpeta ya existe
print("\nEstructura del proyecto generada correctamente.")

# Crear un archivo .gitkeep en cada directorio para guardar la estructura de los folders creados
# desde el primer pull que hagamos al repositorio
for folder in folders:
    directory = PROJECT_ROOT / folder
    directory.mkdir(parents=True, exist_ok=True)

    gitkeep = directory / ".gitkeep"
    gitkeep.touch(exist_ok=True)

    print(f"Marcador creado: {folder}/.gitkeep")

print("\nEstructura preparada para Git.")
````

## Verificación de la estructura del proyecto

Una vez creada la estructura base del proyecto, se realiza una verificación automática de directorios y archivos con el objetivo de comprobar que todas las carpetas se hayan generado correctamente y que la organización del repositorio corresponda a la estructura planificada.

### Recorrido recursivo de la estructura

Para llevar a cabo esta validación, se utiliza el método `rglob("*")`, el cual recorre de forma recursiva todos los elementos contenidos dentro del directorio raíz del proyecto.

Posteriormente, la función `sorted()` permite ordenar los resultados para obtener una visualización consistente y facilitar la inspección de la estructura generada.

### Identificación de directorios y archivos

Durante el recorrido, cada elemento encontrado es clasificado según su tipo:

- Si corresponde a una carpeta, se muestra con la etiqueta `[DIR]`.
- Si corresponde a un archivo, se muestra con la etiqueta `[FILE]`.

Esta diferenciación facilita la lectura de la salida y permite identificar rápidamente la jerarquía de directorios y los archivos presentes dentro del proyecto.

### Uso de rutas relativas

En lugar de mostrar las rutas absolutas del sistema operativo, se utilizan rutas relativas respecto al directorio raíz del proyecto mediante `relative_to(PROJECT_ROOT)`.

Esto mejora la legibilidad de la salida y permite que la estructura sea comprensible independientemente del equipo o sistema operativo donde se ejecute el Notebook.

### Exclusión de archivos temporales de Jupyter

Se excluyen los directorios y archivos asociados a `.ipynb_checkpoints`, los cuales son generados automáticamente por Jupyter Notebook como mecanismo de respaldo.

Dado que estos archivos no forman parte de la estructura funcional del proyecto y pueden introducir ruido en la inspección, se omiten del listado para centrar la verificación únicamente en los elementos relevantes del repositorio.

### Propósito de la validación

Esta comprobación permite:

- Confirmar que todos los directorios planificados fueron creados correctamente.
- Verificar que los archivos auxiliares, como `.gitkeep`, se encuentren en las ubicaciones esperadas.
- Detectar posibles errores de creación o ubicación de carpetas.
- Garantizar que todos los integrantes del equipo trabajen sobre una estructura homogénea del proyecto.

De esta manera, se obtiene una visión completa del estado actual del repositorio antes de iniciar las etapas de adquisición, procesamiento y análisis de los datos.

````python

for path in sorted(PROJECT_ROOT.rglob("*")):
    if path.is_dir():
        print(f"[DIR]  {path.relative_to(PROJECT_ROOT)}")
    elif ".ipynb_checkpoints" not in str(path):
        print(f"[FILE] {path.relative_to(PROJECT_ROOT)}")
````

## Configuración y validación automática del entorno de trabajo

Una vez definida la estructura del proyecto, se procede a preparar el entorno de desarrollo mediante la gestión automatizada de dependencias. Para ello, previamente se completaron tres elementos fundamentales del repositorio:

- **Archivo `requirements.txt`**: contiene el listado de librerías necesarias para reproducir el análisis y ejecutar correctamente todos los componentes del proyecto. Este lo mantendremos en constante actualización, de acuerdo con la necesidad del proyecto.
- **Archivo `README.md`**: incorpora información preliminar sobre los objetivos del proyecto, la organización del repositorio y las instrucciones básicas de uso. El cuál se actualizará hasta concluír el proyecto para incorporar información que sea útil para la consulta del proyecto.
- **Archivo `.gitignore`**: define los archivos y directorios que no deben ser versionados mediante Git, evitando incorporar archivos temporales, configuraciones locales o artefactos generados durante la ejecución.

Esta preparación inicial permite garantizar que todos los integrantes del equipo trabajen bajo las mismas condiciones y utilizando un conjunto consistente de dependencias.

### Localización automática del archivo de dependencias

El procedimiento comienza identificando el directorio de trabajo actual desde el que se está ejecutando el Notebook. A partir de esta ubicación, se realiza una búsqueda ascendente a través de todos los directorios padre hasta localizar el archivo `requirements.txt`.

Esta estrategia aporta flexibilidad al proyecto, ya que permite ejecutar el Notebook desde distintos niveles de la estructura de carpetas sin necesidad de modificar rutas manualmente.

### Instalación automática de dependencias

Una vez localizado el archivo `requirements.txt`, se ejecuta automáticamente el proceso de instalación de paquetes utilizando el mismo intérprete de Python que está ejecutando el Notebook.

Para ello, se invoca internamente el comando:

```bash
python -m pip install -r requirements.txt

````python
ruta_actual = Path.cwd()

requirements = None

# Busca el archivo requirements.txt comenzando por la carpeta actual
# y continuando por todas sus carpetas padre hasta llegar a la raíz
for carpeta in [ruta_actual] + list(ruta_actual.parents):
    archivo = carpeta / "requirements.txt"
    if archivo.exists():
        requirements = archivo
        break

if requirements:
    print(f"Encontrado: {requirements}")
# Ejecuta el equivalente a:
    # python -m pip install -r requirements.txt
    # usando el mismo intérprete que ejecuta el notebook
    subprocess.check_call([
        sys.executable,
        "-m",
        "pip",
        "install",
        "-r",
        str(requirements)
    ])

    print("\nDependencias instaladas:\n")

    with open(requirements, encoding="utf-8") as f:
        for linea in f:
            paquete = linea.strip()

            if not paquete or paquete.startswith("#"):
                continue

            # Elimina restricciones de versión si existen
            nombre = (
                paquete.split("==")[0]
                .split(">=")[0]
                .split("<=")[0]
                .strip()
            )

            try:
                print(f"✓ {nombre}: {version(nombre)}")
            except PackageNotFoundError:
                print(f"✗ {nombre}: no encontrado")

else:
    print("No se encontró requirements.txt")
````
