```python

from pathlib import Path    #Path nos permite trabajar con rutas y directorios.

# Directorio principal del proyecto 
PROJECT_ROOT = Path.cwd()   # Obtiene el directorio de trabajo actual del Notebook

# Recordar que Path.cwd() utiliza el directorio de trabajo del kernel, que no necesariamente deberá
# coincidir con la ubicación física del Notebook. Por lo que se deberá ejecutar el Notebook en la
# raíz del proyecto como directorio de trabajo.

# Estructura de directorios que se desean crear para el proyecto
folders = [
    "data/raw",
    "data/processed",
    "data/external",
    "notebooks",
    "src",
    "reports/figures",
    "reports/outputs",
    "models"
]

# Crear directorios
for folder in folders:
    path = PROJECT_ROOT / folder
    path.mkdir(parents=True, exist_ok=True)    # mkdir() >> Crea el directorio 
    print(f"Directorio creado: {path}")        # parents=True >> Crea los directorios superiores 
                                               # exist_ok=True >> Evita errores si la carpeta ya existe
print("\nEstructura del proyecto generada correctamente.")
```

```python

for path in sorted(PROJECT_ROOT.rglob("*")):
    if path.is_dir():
        print(f"[DIR]  {path.relative_to(PROJECT_ROOT)}")
    elif ".ipynb_checkpoints" not in str(path):
        print(f"[FILE] {path.relative_to(PROJECT_ROOT)}")
```
